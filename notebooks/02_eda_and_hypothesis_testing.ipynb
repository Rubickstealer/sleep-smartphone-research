```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as seaborn_plt  # импортируем для красивых графиков
import os

# Загружаем наши очищенные данные из data/processed/
processed_file_path = os.path.join('..', 'data', 'processed', 'sleep_data_cleaned.csv')
df_clean = pd.read_csv(processed_file_path)

print("Первые 5 строк очищенного датасета:")
display(df_clean.head())

# Базовая статистическая сводка
print("\nСтатистические характеристики данных:")
display(df_clean.describe())

# Считаем коэффициент корреляции между временем в смартфоне перед сном и качеством сна
correlation = df_clean['night_screen_time'].corr(df_clean['sleep_quality_score'])
print(f"\nКоэффициент корреляции между временем в телефоне и качеством сна: {correlation:.2f}")
```

```python
# Настраиваем стиль графиков
seaborn_plt.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
seaborn_plt.scatterplot(
    data=df_clean, 
    x='night_screen_time', 
    y='sleep_quality_score', 
    alpha=0.7, 
    color='purple',
    s=70
)

plt.title('Зависимость качества сна от времени в смартфоне перед сном', fontsize=12)
plt.xlabel('Время в смартфоне перед сном (часы)', fontsize=10)
plt.ylabel('Оценка качества сна (от 1 до 10)', fontsize=10)

plt.show()
```
