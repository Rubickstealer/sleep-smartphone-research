```python
import pandas as pd
import numpy as np
import os

# Убедимся, что папка data/raw существует
os.makedirs('../data/raw', exist_ok=True)

# Генерируем тестовый датасет на 100 строк для нашего исследования
np.random.seed(42)
n_samples = 100

synthetic_data = {
    'user_id': range(1, n_samples + 1),
    'night_screen_time': np.random.uniform(0.5, 6.0, n_samples), # часы в смартфоне перед сном
    'sleep_duration': np.random.normal(7.0, 1.5, n_samples),     # часы сна
    'sleep_quality_score': np.random.randint(1, 11, n_samples)   # оценка качества сна от 1 до 10
}

df_raw = pd.DataFrame(synthetic_data)

# Специально добавим несколько пропусков (NaN), чтобы показать очистку данных
df_raw.loc[5, 'sleep_duration'] = np.nan
df_raw.loc[12, 'sleep_quality_score'] = np.nan
df_raw.loc[20, 'night_screen_time'] = np.nan

# Сохраняем как сырой файл в data/raw/
raw_file_path = '../data/raw/sleep_data.csv'
df_raw.to_csv(raw_file_path, index=False)

print(f"Тестовый файл успешно создан и сохранен по пути: {raw_file_path}")

# Проверяем загрузку
df = pd.read_csv(raw_file_path)
print("\nПервые 5 строк сгенерированных данных:")
display(df.head())
```

```python

print("Пропущенные значения до очистки:")
print(df.isnull().sum())


df_clean = df.dropna().copy()

print(f"\nРазмер данных до очистки: {df.shape}")
print(f"Размер данных после очистки: {df_clean.shape}")


processed_dir = os.path.join('..', 'data', 'processed')
os.makedirs(processed_dir, exist_ok=True)


processed_file_path = os.path.join(processed_dir, 'sleep_data_cleaned.csv')
df_clean.to_csv(processed_file_path, index=False)

print(f"\nОбработанные данные успешно сохранены: {processed_file_path}")
```
